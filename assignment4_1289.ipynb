```python
import matplotlib.pyplot as plt
import networkx as nx
edges = [
    ("T1","T2"),
    ("T1","T3"),
    ("T2","T4"),
    ("T3","T4"),
    ("T4","T5"),
    ("T5","T6"),
    ("T3","T6"),
    ("T6","T7"),
    ("T2","T7"),
]

G = nx.Graph()
G.add_edges_from(edges)

num_nodes = G.number_of_nodes()
num_edges = G.number_of_edges()

print(f"Number of nodes: {num_nodes}")
print(f"Number of edges: {num_edges}\n")

degrees = dict(G.degree())
max_degree = max(degrees.values())
highest_degree_nodes = [node for node, degree in degrees.items() if degree == max_degree]

print(f"Highest degree: {max_degree}")
print(f"Nodes with highest degree: {highest_degree_nodes}\n")

shortest_path_t1_t7 = nx.shortest_path_length(G, source="T1", target="T7")
shortest_path_len = nx.shortest_path_length(G, source="T1", target="T7")
print(f"Shortest path length from T1 to T7: {shortest_path_t1_t7}")
print(f"Shortest path length : {shortest_path_len}\n")

deg_centrality = nx.degree_centrality(G)
bet_centrality = nx.betweenness_centrality(G)
clos_centrality = nx.closeness_centrality(G)

print(f"Degree centrality: {deg_centrality}\n")

is_connected = nx.is_connected(G)
num_components = nx.number_connected_components(G)
print(f"Is the graph connected? {is_connected}\n")
print(f"Number of connected components: {num_components}\n")

max_betweenness_node = max(bet_centrality, key= bet_centrality.get)

node_colors = []
node_sizes = []

for node in G.nodes():
    if node in highest_degree_nodes:
        node_colors.append("crimson")
        node_sizes.append(1200)
    else:
        node_colors.append("skyblue")
        node_sizes.append(700)

pos = nx.spring_layout(G, seed = 42)

plt.figure(figsize=(9,7))

nx.draw_networkx_nodes(
    G, pos,node_color=node_colors, node_size=node_sizes, edgecolors="black"
)

nx.draw_networkx_edges(G, pos, width = 2,alpha=0.8,  edge_color="darkblue") # Increased alpha and changed color for better visibility

nx.draw_networkx_labels(
    G,pos, font_size=11, font_family="sans-serif", font_weight = "bold", alpha = 0.7, font_color= "black"
)

plt.title("Telecom Tower Network Connectivity", fontsize=14, fontweight = "bold")
plt.axis("off")

plt.plot(
    [],
    [],
    marker= "o",
    color= "crimson",
    linestyle= "None",
    markersize= 12,
    label= f"Highest Betweenness ({max_betweenness_node})",
)
plt.plot(
    [],
    [],
    marker= "o",
    color= "skyblue",
    linestyle= "None",
    markersize= 9,
    label= "Standard Towers",
)
plt.legend(loc="lower right", frameon=True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import networkx as nx

edges =[
    ("Amanda", "Bobby"),
    ("Amanda", "CocaCola"),
    ("Bobby", "Daud"),
    ("CocaCola", "Daud"),
    ("Daud","Escalator"),
    ("Escalator","Fukre"),
    ("ColaCola","Fukre"),
]

G = nx.Graph()
G.add_edges_from(edges)

num_nodes = G.number_of_nodes()
num_nodes_edges = G.number_of_edges() # Renamed to avoid confusion with num_nodes

print(f"Number of nodes (students): {num_nodes}")
print(f"Number of edges (friendship): {num_nodes_edges}\n") # Used num_nodes_edges

degrees = dict(G.degree())
for student, deg in degrees.items():
  print(f"Degree of {student}: {deg}")
print()

max_degree_val = max (degrees.values())
highest_degree_students = [student for student, deg in degrees.items() if deg == max_degree_val]
print(f"Highest degree: {max_degree_val}) : {','.join(highest_degree_students)}\n")

has_direct_friendship = G.has_edge("Amanda", "Escalator") # Corrected typo in 'Escalotor'
print(f"Does Amanda and Escalator have a direct friendship? {has_direct_friendship}\n") # Corrected typo in 'Escalotor'

shortest_path = nx.shortest_path(G, source="Amanda", target="Fukre")
shortest_path_len = nx.shortest_path_length(G, source="Amanda", target="Fukre")
print(f"Shortest path from Amanda to Fukre: {shortest_path}")
print(f"Shortest path length: {shortest_path_len}\n")

pos = nx.spring_layout(G, seed=42)

plt.figure(figsize=(8,6))

node_colors = [
    "orange" if node in highest_degree_students else "lightgreen"
    for node in G.nodes()
]

nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=1000, edgecolors="black")
nx.draw_networkx_edges(G, pos, width=2, alpha=0.7, edge_color="darkblue")

nx.draw_networkx_labels(G,pos,font_size=10, font_family="sans-serif", font_weight="bold") # Corrected function name

plt.title("School friendship network", fontsize=14, fontweight="bold")
plt.axis("off")

plt.plot(
    [],
    [],
    marker="o",
    color="orange",
    linestyle="None",
    markersize=10,
    label=f"most connected students", # Corrected typo
)

plt.plot(
    [],
    [],
    marker="o",
    color="lightgreen",
    linestyle="None",
    markersize=8,
    label="standard students",
)

plt.legend(loc="upper right", frameon=True)

plt.tight_layout()
plt.show()
```
